CARTEGORICAL SUMMARY

```python
print(df['gender'].value_counts())
print(df['cartegories'].value_counts())
print(df['timeslot'].value_counts())
print(df['branch'].value_counts())
print(df['city'].value_counts())
print(df['customer type'].value_counts())
print(df['payment'].value_counts())
print(df['sales month'].value_counts())
print(df['sales year'].value_counts())
```

1. Customer Demographics

The customer base is evenly distributed by gender:
Female: 50.1%
Male: 49.9%
There is no gender dominance in purchasing behavior, indicating that products and marketing strategies currently appeal equally to both segments. This presents an opportunity to maintain balanced targeting rather than prioritizing one gender over the other.

2. Product Category Performance

Sales are relatively evenly spread across all categories, with slight variations:
Fashion Accessories (17.8%)
Food & Beverages (17.4%)
Electronic Accessories (17.0%)
Lower Performing Categories:
Sports & Travel (16.6%)
Home & Lifestyle (16.0%)
Health & Beauty (15.2%)
No single category dominates, suggesting a diversified revenue stream. However, Fashion accessories and food-related products slightly outperform others. Health & beauty shows the lowest share, indicating a potential area for growth or review.

3. Peak Sales Periods (Time of Day)
Afternoon: 45.4% (Peak period)
Evening: 35.5%
Morning: 19.1%

The afternoon period is the primary revenue driver, accounting for nearly half of all transactions.
Morning traffic is significantly lower, suggesting underutilized capacity. We optimize staffing, promotions, and inventory availability for afternoon peaks. Introduce targeted incentives (e.g., discounts) to boost morning sales.

4. Branch Performance
Branch A: 34.0%
Branch B: 33.2%
Branch C: 32.8%
Operational performance across locations is highly consistent, indicating Standardized processes are effective No immediate performance gaps between branches

5. Geographic Distribution (City-Level)
Los Angeles: 241 transactions (highest)
San Francisco: 163 transactions
Other cities contribute smaller shares

Los Angeles is the primary market, driving the largest portion of sales.
There is a noticeable drop-off in activity outside the top city, suggesting:
Strong regional concentration
Potential to expand or strengthen presence in lower-performing cities

Below is the visual for easy understanding

```python
import matplotlib.pyplot as plt
plt.rcParams["font.family"] = "DejaVu Sans"
```

GENDER DUSTRIBUTION CHART

```python
gender_counts = df['gender'].value_counts()

plt.figure()
gender_counts.plot(kind='bar')
plt.title('Customer Distribution by Gender')
plt.xlabel('gender')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

SALES DISTRIBUTION BY PRODUCTS

```python
category_counts = df['cartegories'].value_counts()
plt.figure()
category_counts.plot(kind='bar')
plt.title('Sales Distribution by Product Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

SALES DISTRIBUTION BY TIME SLOT

```python
timeslot_counts = df['timeslot'].value_counts()

plt.figure()
timeslot_counts.plot(kind='bar')
plt.title('Sales by Time of Day')
plt.xlabel('Time Slot')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

BRANCH PERFORMNCE

```python
branch_counts = df['branch'].value_counts()

plt.figure()
branch_counts.plot(kind='bar')
plt.title('Sales by Branch')
plt.xlabel('Branch')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

CITY PERFORMANCE

```python
city_counts = df['city'].value_counts().head(20)

plt.figure()
city_counts.plot(kind='barh')
plt.title('Sales by City')
plt.xlabel('City')
plt.ylabel('Count')
plt.xticks(rotation=60)
plt.show()
```

YEARLY PERFORMANCE

```python

year_counts = df['sales year'].value_counts().sort_index()

plt.figure()
year_counts.plot(kind='line', marker='o')
plt.title('Sales Trend Over Years')
plt.xlabel('Year')
plt.ylabel('Count')
plt.show()
```
